## What are FLOPs?

Recall that each matrix multiplication in a transformer block is composed of multiplication and addition operations. Each operation is a **floating point operation**, or **FLOP**, since it involves calculations between floating point numbers.

To measure the total computational cost of a task, such as training a model, you can count up all of these individual operations. The total count is referred to as the number of **FLOPs**. In the example provided in the previous article "What happens in the GPU during training and during inference?", a simple matrix multiplication between a $2 \times 3$ matrix and a $3 \times 2$ matrix required 20 operations. However, training a language model can require billions, trillions or even more of these operations.

------
> **ℹ️ Info: FLOPs versus FLOPS:**
>
>You will often see two very similar acronyms: FLOPs and FLOPS. The capitalization of the final S is crucial as it signals a different meaning.
>
>1. **FLOPs** (Floating Point Operations) is the plural of **FLOP**. It represents the total number of operations required to perform a specific training process or perform inference on a specific example.
>
>   * For example: "Training this model will require 500 trillion FLOPs."
>
>2. **FLOPS** (Floating Point Operations Per Second) is a rate of measurement. It is a measure of computational performance, or speed. It tells you how many floating point operations (FLOPs) a piece of specific hardware (e.g., a GPU) can perform per second. The capitalized S therefore stands for "per Second."
>
>   * For example: "This GPU has a performance of 200 TeraFLOPS ($200 \times 10^{12}$ FLOPS)."
>
>To see the connection between FLOPs and FLOPS, consider a model training process that requires 1,000 FLOPs on a GPU that processes at 100 FLOPS. In this case, it would take 10 seconds to train the model, since the GPU can perform 100 FLOPs per second (i.e., 100 FLOPS).
>
------

## Estimating FLOPs

A widely used "back-of-the-envelope" formula for estimating the number of FLOPs required for one epoch of training (i.e., one full pass over the dataset) is:

$$\text{FLOPs} \approx 6 \times P \times N$$


where $P$ is the total number of trainable model parameters and $N$ is the total number of tokens in the training dataset.

The scaling factor of 6 is a widely used rule of thumb that accounts for the work done during one training step:

* **Forward pass**: The cost is approximately $2 \times \text{P} \times \text{N}$ FLOPs. The factor of 2 comes from the fact that, for each parameter, there is roughly one multiplication and one addition (a "multiply-accumulate" operation).

* **Backward pass**: The cost is approximately $4 \times \text{P} \times \text{N}$ FLOPs. This is because the backward pass, which involves calculating gradients, is roughly twice as computationally expensive as the forward pass [1].

Adding these two components together gives the final estimation of $6 \times P \times N$.

You can read more about methods for estimating FLOPs in the paper [Training Compute-Optimal Large Language Models](https://arxiv.org/abs/2203.15556) from Google DeepMind [2].


## Coding Activity 1: Implement the FLOPs estimation formula

In preparation for computing FLOPs, you need a function that returns the number of FLOPs.

------
> 💻 **Your task**:
>
> Complete the function `compute_num_flops(param_count: float, num_tokens: float)`.
>
> It should return a single number representing the number of FLOPs according to the estimation formula.
>
> Once you have implemented this function, run both cells to define the function and test your code.
------

```python
def compute_num_flops(param_count: float, num_tokens: float) -> float:
    """Estimates the training FLOPs for one epoch.

    Args:
      param_count: The total number of trainable parameters in the model.
      num_tokens: The total number of tokens in the training dataset.

    Returns:
      The estimated total number of floating point operations.
    """
    total_flops = 6 * param_count * num_tokens

    return total_flops
```

### Thí nghiệm kiểm chứng Linear Scaling Law (Task 3)

Thay vì kéo slider thủ công, cell dưới đây tự động hoá đúng thí nghiệm được yêu cầu:

* **Cố định** `log_num_tokens = 10` (tức $N = 10^{10}$ token).
* **Tăng** `log_param_count` từ 8 lên 12, mỗi bước 1 bậc (tức $P$ tăng 10× mỗi bước).
* Quan sát cột *Hệ số tăng so với dòng trên*.

```python
# @title Kiểm chứng: tăng P lên 10 lần thì FLOPs tăng đúng 10 lần
# @markdown Cố định N = 10^10 token, tăng P từ 10^8 -> 10^12.

log_num_tokens_fixed = 10  # @param {type:"slider", min:9, max:14, step:1}

num_tokens_fixed = 10**log_num_tokens_fixed
scaling_rows = []
previous_flops = None

for log_param_count in range(8, 13):
    param_count = 10**log_param_count
    total_flops = compute_num_flops(param_count, num_tokens_fixed)

    if previous_flops is None:
        growth = "—"
    else:
        growth = f"x{total_flops / previous_flops:.1f}"

    scaling_rows.append({
        "log_param_count": log_param_count,
        "Parameters (P)": f"1e{log_param_count}",
        "Tokens (N)": f"1e{log_num_tokens_fixed}",
        "Training FLOPs": f"{total_flops:.2e}",
        "Hệ số tăng so với dòng trên": growth,
    })
    previous_flops = total_flops

display(pd.DataFrame(scaling_rows))
```

**Quan sát:** khi `log_param_count` đi từ 8 → 9 (với $N = 10^{10}$ cố định), FLOPs đi từ
$6.00 \times 10^{18}$ → $6.00 \times 10^{19}$: **phần định trị (6.00) không đổi, chỉ số mũ
tăng đúng 1 đơn vị**, tương ứng đúng 10×.

**Kết luận (Linear Scaling Law):** trong công thức $\text{FLOPs} = 6PN$, nhân $P$ (hoặc $N$)
với $k$ sẽ nhân tổng FLOPs với đúng $k$. Trên đồ thị log–log, đây là một đường thẳng có
**độ dốc bằng 1** — chính là dạng biểu đồ mà mọi bài báo Scaling Laws
(Kaplan et al. 2020, Hoffmann et al. 2022) sử dụng.

## Task 4: Ứng dụng thực tế — Quy đổi FLOPs sang GPU Hours & Chi phí

FLOPs mới chỉ là "khối lượng công việc". Để trả lời câu hỏi của lãnh đạo — *"train bao lâu,
tốn bao nhiêu tiền?"* — cần quy đổi qua năng lực phần cứng:

$$\text{Training Time (giây)} = \frac{\text{Total Training FLOPs}}{\text{Số GPU} \times \text{Peak FLOPS} \times \text{MFU}}$$

$$\text{GPU Hours} = \text{Training Time (giờ)} \times \text{Số GPU}
\qquad\qquad
\text{Cost} = \text{GPU Hours} \times \text{Đơn giá thuê / GPU-giờ}$$

Trong đó:

* **Peak FLOPS** — tốc độ lý thuyết nhà sản xuất công bố. NVIDIA A100 SXM4 80GB đạt
  $\approx 312$ TFLOPS $= 312 \times 10^{12}$ FLOPS ở FP16/BF16 Tensor Core.
* **MFU (Model FLOPs Utilization)** — hiệu suất thực tế sau khi trừ hao truyền dữ liệu qua
  mạng, nạp bộ nhớ, đồng bộ gradient. Thực tế thường đạt **0.30 – 0.45**.

**Tình huống:** huấn luyện mô hình **7B tham số** trên **2T token**, thuê **64 GPU A100**,
**MFU = 0.35**, giá **\$2.50 / GPU-giờ**.

```python
# Hằng số phần cứng / giá thuê
A100_PEAK_FLOPS = 312e12   # 312 TFLOPS ở FP16/BF16 Tensor Core
PRICE_PER_GPU_HOUR = 2.50  # USD cho 1 GPU A100 trong 1 giờ

SECONDS_PER_HOUR = 3600
HOURS_PER_DAY = 24


def estimate_training_budget(
    param_count: float,
    num_tokens: float,
    num_gpus: int,
    peak_flops_per_gpu: float,
    mfu: float,
    price_per_gpu_hour: float,
) -> dict:
    """Quy đổi từ quy mô mô hình/dữ liệu sang thời gian và chi phí huấn luyện.

    Args:
      param_count: Tổng số tham số huấn luyện được của mô hình.
      num_tokens: Tổng số token trong tập dữ liệu huấn luyện.
      num_gpus: Số lượng GPU trong cụm.
      peak_flops_per_gpu: Peak FLOPS lý thuyết của 1 GPU.
      mfu: Model FLOPs Utilization, trong khoảng (0, 1].
      price_per_gpu_hour: Đơn giá thuê 1 GPU trong 1 giờ (USD).

    Returns:
      Dictionary chứa FLOPs, năng lực thực tế của cụm, thời gian, GPU hours và chi phí.
    """
    # Bước 1: tổng FLOPs lý thuyết cho 1 epoch.
    total_flops = compute_num_flops(param_count, num_tokens)

    # Bước 2: năng lực tính toán thực tế của cả cụm (FLOPS).
    effective_flops = num_gpus * peak_flops_per_gpu * mfu

    training_seconds = total_flops / effective_flops
    wall_clock_hours = training_seconds / SECONDS_PER_HOUR
    gpu_hours = wall_clock_hours * num_gpus

    # Bước 3: chi phí thuê cloud.
    total_cost_usd = gpu_hours * price_per_gpu_hour

    return {
        "total_flops": total_flops,
        "effective_flops": effective_flops,
        "training_seconds": training_seconds,
        "wall_clock_hours": wall_clock_hours,
        "wall_clock_days": wall_clock_hours / HOURS_PER_DAY,
        "gpu_hours": gpu_hours,
        "total_cost_usd": total_cost_usd,
    }


budget = estimate_training_budget(
    param_count=7e9,          # 7B tham số
    num_tokens=2e12,          # 2T token
    num_gpus=64,
    peak_flops_per_gpu=A100_PEAK_FLOPS,
    mfu=0.35,
    price_per_gpu_hour=PRICE_PER_GPU_HOUR,
)

print("Bước 1 — Tổng FLOPs (6 x P x N):")
print(f"    6 x 7e9 x 2e12 = {budget['total_flops']:.2e} FLOPs\n")

print("Bước 2 — Năng lực thực tế của cụm và thời gian huấn luyện:")
print(f"    Effective FLOPS = 64 x 312e12 x 0.35 = {budget['effective_flops']:.4e} FLOPS")
print(f"    Thời gian       = {budget['training_seconds']:,.0f} giây")
print(f"                    = {budget['wall_clock_hours']:,.1f} giờ (wall-clock)")
print(f"                    = {budget['wall_clock_days']:,.1f} ngày")
print(f"    GPU Hours       = {budget['wall_clock_hours']:,.1f} x 64 = {budget['gpu_hours']:,.0f} GPU-giờ\n")

print("Bước 3 — Chi phí thuê cloud:")
print(f"    {budget['gpu_hours']:,.0f} GPU-giờ x ${PRICE_PER_GPU_HOUR:.2f} = ${budget['total_cost_usd']:,.2f}")
```

### Phân tích độ nhạy: MFU và quy mô cụm ảnh hưởng thế nào?

Hai "cần gạt" mà đội kỹ thuật có thể tác động là **số GPU** (thuê thêm máy) và **MFU**
(tối ưu code, batch size, communication overlap). Bảng dưới so sánh các kịch bản.

```python
sensitivity_rows = []

for mfu in (0.30, 0.35, 0.45):
    for num_gpus in (64, 128, 256):
        scenario = estimate_training_budget(
            param_count=7e9,
            num_tokens=2e12,
            num_gpus=num_gpus,
            peak_flops_per_gpu=A100_PEAK_FLOPS,
            mfu=mfu,
            price_per_gpu_hour=PRICE_PER_GPU_HOUR,
        )
        sensitivity_rows.append({
            "MFU": f"{mfu:.0%}",
            "Số GPU": num_gpus,
            "Thời gian (ngày)": f"{scenario['wall_clock_days']:.1f}",
            "GPU Hours": f"{scenario['gpu_hours']:,.0f}",
            "Chi phí (USD)": f"${scenario['total_cost_usd']:,.0f}",
        })

display(pd.DataFrame(sensitivity_rows))
```

   từ 139 ngày xuống ~35 ngày, nhưng tổng GPU-hours và chi phí **không đổi** (213,675 GPU-giờ,

2. **MFU mới là đòn bẩy chi phí thật sự.** Cải thiện MFU từ 0.30 lên 0.45 (tối ưu kernel, batch
   size, overlap communication) cắt chi phí từ \$623,219 xuống \$415,480 — **tiết kiệm ~\$208,000
   (33%)** mà không cần thuê thêm một GPU nào.
3. **Con số 139 ngày là cảnh báo rủi ro vận hành.** Một lần train kéo dài 4.5 tháng trên 64 GPU
   gần như chắc chắn sẽ gặp lỗi node giữa chừng — nên phần dự toán phải cộng thêm ngân sách cho
   checkpointing và cho các lần restart, đồng thời cân nhắc cụm lớn hơn để rút ngắn cửa sổ rủi ro.

```python
# This is a complete implementation of `compute_num_flops`.
def compute_num_flops(param_count, num_tokens) -> float:
    """Estimates the training FLOPs for one epoch.

    Args:
      param_count: The total number of trainable parameters in the model.
      num_tokens: The total number of tokens in the training dataset.

    Returns:
      The estimated total number of floating point operations.
    """

    total_flops = 6 * param_count * num_tokens

    return total_flops
```
